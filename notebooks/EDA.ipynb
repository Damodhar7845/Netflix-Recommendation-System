```python
import pandas as pd

ratings = []

movie_id = None
movie_count = 0

with open("../data/combined_data_1.txt", "r") as f:

    for line in f:

        line = line.strip()

        if ":" in line:
            movie_id = int(line.replace(":", ""))
            movie_count += 1

            if movie_count > 500:
                break

        else:
            user_id, rating, date = line.split(",")

            ratings.append([
                int(user_id),
                movie_id,
                int(rating)
            ])

ratings_large = pd.DataFrame(
    ratings,
    columns=["UserID", "MovieID", "Rating"]
)

ratings_large.head()
```

```python
movie_ratings = ratings_with_titles.groupby('Title')['Rating'].agg(
    ['count', 'mean']
).reset_index()

movie_ratings.head()
```

```python
movie_ratings = movie_ratings.rename(
    columns={
        'count': 'num_ratings',
        'mean': 'avg_rating'
    }
)

movie_ratings.head()
```

```python
movie_ratings.sort_values(
    by='num_ratings',
    ascending=False
).head(10)
```

```python
popular_movies = movie_ratings[
    movie_ratings['num_ratings'] >= 500
]

popular_movies.shape
```

```python
popular_df = ratings_with_titles.merge(
    popular_movies[['Title']],
    on='Title'
)

popular_df.shape
```

```python
user_movie_matrix = popular_df.pivot_table(
    index='Title',
    columns='UserID',
    values='Rating'
)

user_movie_matrix.shape
```

```python
user_movie_matrix = user_movie_matrix.fillna(0)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_scores = cosine_similarity(
    user_movie_matrix
)

similarity_scores.shape
```

```python
def recommend(movie_name):

    idx = user_movie_matrix.index.get_loc(movie_name)

    distances = similarity_scores[idx]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for movie in movies_list:
        print(user_movie_matrix.index[movie[0]])
```

```python
movie_info = ratings_with_titles.groupby('Title').agg(
    avg_rating=('Rating', 'mean'),
    num_ratings=('Rating', 'count')
).reset_index()

movie_info.head()
```

```python
def recommend(movie_name):

    # Check if movie exists
    if movie_name not in user_movie_matrix.index:
        return pd.DataFrame({
            "Error": ["Movie not found in dataset"]
        })

    idx = user_movie_matrix.index.get_loc(movie_name)

    distances = similarity_scores[idx]

    movies_list = sorted(
        list(enumerate(distances)),
        key=lambda x: x[1],
        reverse=True
    )[1:6]

    recommended_movies = []

    for movie in movies_list:

        title = user_movie_matrix.index[movie[0]]

        avg_rating = movie_info.loc[
            movie_info['Title'] == title,
            'avg_rating'
        ].values[0]

        num_ratings = movie_info.loc[
            movie_info['Title'] == title,
            'num_ratings'
        ].values[0]

        similarity = round(movie[1] * 100, 2)

        recommended_movies.append([
            title,
            round(avg_rating, 2),
            num_ratings,
            similarity
        ])

    return pd.DataFrame(
        recommended_movies,
        columns=[
            "Movie",
            "Avg Rating",
            "Number of Ratings",
            "Similarity (%)"
        ]
    )
```
